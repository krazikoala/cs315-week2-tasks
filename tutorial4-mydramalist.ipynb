## AGGREGATE PAGE

```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
headers = {"User-Agent": "Mozilla/5.0"}
r = requests.get("https://mydramalist.com/shows/top", headers=headers)
soup = BeautifulSoup(r.content, "html.parser")

for box in soup.find_all("div", class_="box"):
    title = box.find("h6", class_="title")
    if title:
        print(title.find("a").text.strip())
```

```python
#Results of the scrape

BASE = "https://mydramalist.com"

for h6 in soup.find_all("h6", class_="title"):
    link = h6.find("a")
    title = link.text.strip()
    url = BASE + link["href"]
    print(title, "|", url)
```

```python
h6 = soup.find("h6", class_="title")
card = h6.find_parent("div", class_="box")
print(card.prettify()[:1500])
```
