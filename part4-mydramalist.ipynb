## AGGREGATE PAGE

```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
URL = "https://mydramalist.com/shows/top"
page = requests.get(URL, headers={"User-Agent": "Mozilla/5.0"})
soup = BeautifulSoup(page.content, "html.parser")

titles = soup.find_all("h6", class_="title")

for title_element in titles:
    print(title_element.find("a").text.strip())
```

```python
for box in soup.find_all("div", class_="box"):
    title = box.find("h6", class_="title")
    if not title:
        continue
    name=title.find("a").text.strip()
    meta=box.find("span", class_="text-muted")
    score=box.find("span", class_="score")
    print(name)
    print(meta.text.strip() if meta else "")
    print(score.text.strip() if score else "")
    print()
```
