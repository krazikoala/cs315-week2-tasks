```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
top_page = requests.get(
    "https://mydramalist.com/shows/top",
    headers={"User-Agent": "Mozilla/5.0"},)
top_soup = BeautifulSoup(top_page.content, "html.parser")
print(top_soup.find("title").text)
print()
box = top_soup.find("h6", class_="title").find_parent("div", class_="box")
print(box.prettify())
```

```python
URL="https://mydramalist.com/shows/top"
page = requests.get(URL, headers={"User-Agent": "Mozilla/5.0"})
soup = BeautifulSoup(page.content, "html.parser")

titles = soup.find_all("h6", class_="title")

for title_element in titles:
    print(title_element.find("a").text.strip())
```

```python
for box in soup.find_all("div", class_="box"):
    title = box.find("h6", class_="title")
    if not title:
        continue
    name=title.find("a").text.strip()
    meta=box.find("span", class_="text-muted")
    score=box.find("span", class_="score")
    print(name)
    if meta:
        print(meta.text.strip())
    else:
        print("")
    if score:
        print(score.text.strip())
    else:
        print("")
    print()
```

```python
#scrape more pages
all_dramas = []
for page_num in range(1, 6):
    URL = "https://mydramalist.com/shows/top?page=" + str(page_num)
    page = requests.get(URL, headers={"User-Agent": "Mozilla/5.0"})
    soup = BeautifulSoup(page.content, "html.parser")

    for box in soup.find_all("div", class_="box"):
        title = box.find("h6", class_="title")
        if not title:
            continue
        name = title.find("a").text.strip()
        meta = box.find("span", class_="text-muted")
        score = box.find("span", class_="score")

        if meta:
            meta_text=meta.text.strip()
        else:
            meta_text=""

        if score:
            score_text=score.text.strip()
        else:
            score_text = ""

        all_dramas.append([name, meta_text, score_text])
```

```python
import csv

with open("top_dramas.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["title", "info", "score"])
    writer.writerows(all_dramas)
```
