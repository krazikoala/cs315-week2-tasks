## AGGREGATE PAGE

```python
import requests
from bs4 import BeautifulSoup
import csv
```

```python
URL="https://mydramalist.com/shows/top"
page=requests.get(URL, headers={"User-Agent": "Mozilla/5.0"})
soup=BeautifulSoup(page.content, "html.parser")

print(soup.find("title").text)

title= soup.find("h6", class_="title")
box= title.find_parent("div", class_="box")
print(box.prettify())
#making it aligned better
```

```python
titles=soup.find_all("h6", class_="title")
for title in titles:
    print(title.find("a").text.strip())
```

```python
for title in titles:
    box=title.find_parent("div", class_="box")
    info=box.find("span", class_="text-muted")
    score=box.find("span", class_="score")
    print(title.find("a").text.strip())
    print(info.text.strip())
    print(score.text.strip())
    print()
```

```python
#scrape more pages and store
file = open("top_dramas.csv", "w")
writer = csv.writer(file)
writer.writerow(["title", "info", "score"])

for page_num in range(1, 6):
    URL = "https://mydramalist.com/shows/top?page=" + str(page_num)
    page = requests.get(URL, headers={"User-Agent": "Mozilla/5.0"})
    soup = BeautifulSoup(page.content, "html.parser")

    titles = soup.find_all("h6", class_="title")
    for title in titles:
        box = title.find_parent("div", class_="box")
        name = title.find("a").text.strip()
        info = box.find("span", class_="text-muted").text.strip()
        score = box.find("span", class_="score").text.strip()
        writer.writerow([name, info, score])

file.close()
```
